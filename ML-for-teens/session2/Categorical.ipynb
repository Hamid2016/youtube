#### only ordinal

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder, OneHotEncoder

# Sample data
df = pd.DataFrame({
    'Education_Level': ['Bachelor', 'Master', 'PhD', 'High School', 'Master']
})

# 1️⃣ Ordinal Encoding (custom order)-recommanded
ordinal_order = [['High School', 'Bachelor', 'Master', 'PhD']]
ordinal_encoder = OrdinalEncoder(categories=ordinal_order)
df['Ordinal_Encoded'] = ordinal_encoder.fit_transform(df[['Education_Level']])

# 2️⃣ Label Encoding (no order, just unique integers)
label_encoder = LabelEncoder()
df['Label_Encoded'] = label_encoder.fit_transform(df['Education_Level'])


# Show result
print(df)
```

#### only nominal

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Sample data
df = pd.DataFrame({
    'Department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# 1️⃣ Dummy Encoding with pandas, drop one column to avoid multicollinearity in linear models.
# Multicollinearity happens when two or more input features are highly correlated
df_dummy = pd.get_dummies(df, columns=['Department'], drop_first=True)

# 2️⃣ One-Hot Encoding with scikit-learn, keep all columns independently. recommanded method
onehot_encoder = OneHotEncoder(sparse_output=False)
onehot_encoded = onehot_encoder.fit_transform(df[['Department']])
onehot_df = pd.DataFrame(onehot_encoded, columns=onehot_encoder.get_feature_names_out(['Department']))

# ✅ Combine original with both encoded versions
df_combined = pd.concat([df, df_dummy, onehot_df], axis=1)


# Show result
print(pd.concat([df, onehot_df], axis=1))

print( pd.concat([df, df_dummy], axis=1))
```
