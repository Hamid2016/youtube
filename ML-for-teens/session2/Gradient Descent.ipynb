#### plotting x vs y and y predicted

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Your data
x = np.array([1, 2, 3, 4, 5], dtype=float).reshape(-1, 1)
y = np.array([2, 4, 6, 8.5, 9.75], dtype=float)

# Fit the model
model = LinearRegression()
model.fit(x, y)

# Extract parameters
slope = model.coef_[0]
intercept = model.intercept_
print(f"Slope: {slope:.4f}")
print(f"Intercept: {intercept:.4f}")

# Predict values
y_pred = model.predict(x)

# Plot
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='blue', label='Original Data')
plt.plot(x, y_pred, color='red', label=f'Linear Fit: y = {slope:.2f}x + {intercept:.2f}')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression on Custom Data')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Define your data
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 6, 8.5, 9.75], dtype=float)

# Step 2: Initialize parameters
m = 0.0
b = 0.0

# Step 3: Set hyperparameters
learning_rate = 0.01
max_iterations = 10000
epsilon = 1e-6

# Step 4: Define the loss function
def compute_loss(x, y, m, b):
    predictions = m * x + b
    return np.mean((predictions - y) ** 2)

# Step 5: Gradient Descent Loop
previous_loss = float('inf')
m_history = []
b_history = []
loss_history = []

for i in range(max_iterations):
    predictions = m * x + b
    error = predictions - y

    dm = np.mean(error * x)
    db = np.mean(error)

    m -= learning_rate * dm
    b -= learning_rate * db

    loss = compute_loss(x, y, m, b)

    m_history.append(m)
    b_history.append(b)
    loss_history.append(loss)

    if abs(previous_loss - loss) < epsilon:
        print(f"Converged at iteration {i}")
        break

    previous_loss = loss

# Final output
print(f"Final parameters: m = {m:.4f}, b = {b:.4f}")
print(f"Final loss: {loss:.6f}")

# Step 6: Plotting m and b over iterations
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(m_history, label='Slope (m)', color='green')
plt.xlabel('Iteration')
plt.ylabel('Value of m')
plt.title('Slope over Iterations')
plt.grid(True)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(b_history, label='Intercept (b)', color='purple')
plt.xlabel('Iteration')
plt.ylabel('Value of b')
plt.title('Intercept over Iterations')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
