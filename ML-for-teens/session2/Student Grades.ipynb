```python
import pandas as pd
```

```python
df=pd.read_csv("grade data/student_grades_modified.csv")
df.head(5)
```

```python
#  returns True when there is at least one missing value
df.isnull().values.any()
```

```python
df.isnull().sum()
```

```python
# Show rows with any missing values
missing_rows = df[df.isnull().any(axis=1)]
print(missing_rows)
```

#### 1.Drop Missing Data

```python
dg=df.dropna()  # Removes rows with any missing values
dg
```

#### 2.Fill with Statistical Measures

```python
# Option 1: Fill with a Placeholder (for categorical)
df['Grade'].fillna('o', inplace=True)

```

```python
# Option 2: Fill with Mode (most frequent value)
df['Student ID'].fillna(df['Student ID'].mode()[0], inplace=True)
```
