#### Read the Data

```python
import pandas as pd

df = pd.read_csv("home_prices.csv", encoding="utf-8")
print(df.head())
```

#### Explore the Data

```python
print(df.head())         # View the first 5 rows
print(df.info())         # Check data types and nulls
print(df.describe())     # Get summary stats
```

#### Visualize the data

```python
import matplotlib.pyplot as plt

plt.scatter(df["Home Area (m2)"], df["Home Price (K€)"], color='blue')
plt.xlabel("Home Area (m2)")
plt.ylabel("Home Price (K€)")
plt.title("Home Price vs Area")
plt.grid(True)
plt.show()
```

#### Fit a Regression Model

```python
from sklearn.linear_model import LinearRegression

X = df[["Home Area (m2)"]]
y = df["Home Price (K€)"]

model = LinearRegression()
model.fit(X, y)

# Predict prices
y_pred = model.predict(X)

# see coefficient and coefficient
print(model.intercept_)
print(model.coef_)

# Plot actual vs predicted
plt.scatter(X, y, color='blue', label='Actual')
plt.plot(X, y_pred, color='red', label='Predicted')
plt.xlabel("Home Area (m2)")
plt.ylabel("Home Price (K€)")
plt.title("Linear Regression Fit")
plt.legend()
plt.show()
```

#### home price equation

```python
#Home Price (K€) = 4.8571 + 2.9486 × Home Area (m²)
```

```python
# Get R² score
score = model.score(X, y)
print("R² Score:", score)
```
