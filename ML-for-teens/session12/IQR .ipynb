```python
import pandas as pd
import numpy as np

# Load dataset (after downloading from Kaggle)
df = pd.read_csv('DataSets/AB_NYC_2019.csv')
df
```

```python
# Check for missing values in each column
missing = df.isnull().sum()
missing
```

```python
# Drop all rows with any missing values
df = df.dropna()

# Check shape after dropping
print("Remaining rows:", df.shape[0])
```

```python
df.isnull().sum()
```

#### Detect outliers in the price column using IQR

```python
# Calculate percentiles
Q1 = df['price'].quantile(0.25)
Q2 = df['price'].quantile(0.50)  # Median
Q3 = df['price'].quantile(0.75)
```

#### Calculate Q1, Q3, and IQR

```python
IQR = Q3 - Q1
IQR
```

```python
# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
# Flag outliers
df['is_outlier'] = (df['price'] < lower_bound) | (df['price'] > upper_bound)
df
```

#### count outliers

```python
# Count outliers
print("Number of outliers:", df['is_outlier'].sum())

# View outlier rows
outliers = df[df['is_outlier']]
```

#### remove outliers

```python
clean_df = df[df['is_outlier'] == False]
clean_df
```
