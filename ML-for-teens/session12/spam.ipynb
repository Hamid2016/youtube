```python
import pandas as pd
```

```python
#read data
df=pd.read_csv("DataSets/spam.csv",encoding="latin-1")
df
```

```python
# Keep only the useful columns
df = df[['v1', 'v2']]
df
```

```python
# Rename columns
df = df.rename(columns={'v1': 'label', 'v2': 'message'})
df
```

```python
# Convert labels to numeric
df['label_num'] = df['label'].map({'ham':0, 'spam':1})
df
```

```python
# Drop the 'label' column
df = df.drop(columns=['label'])
df
```

#### count vectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer

# Initialize CountVectorizer
vectorizer = CountVectorizer()
```

```python
# Fit and transform the message column
X = vectorizer.fit_transform(df['message'])
```

```python
vocab = vectorizer.get_feature_names_out()
vocab
```

```python
# Show up to 150 columns
pd.set_option('display.max_columns', 150)

# Convert sparse matrix to DataFrame
X_df = pd.DataFrame(X.toarray(), columns=vocab)
X_df
```

```python
y=df["label_num"]
y
```

#### Split into train and test sets

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
```

#### Define models + parameter grids

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC

# Naive Bayes
nb_model = MultinomialNB()
nb_params = {'alpha':[0.05,0.1,0.5,1.0]}

# Logistic Regression
lr_model = LogisticRegression(max_iter=2000, n_jobs=-1)
lr_params = {'C':[0.25,0.5,1,2,4], 'solver':['liblinear','saga'], 'penalty':['l2']}

# SVM
svm_model = LinearSVC()
svm_params = {'C':[0.25,0.5,1,2,4], 'loss':['hinge','squared_hinge']}
```

```python
import warnings
warnings.filterwarnings("ignore")
```

#### GridSearchCV

```python
# Naive Bayes
nb_grid = GridSearchCV(nb_model, nb_params, cv=5, scoring='accuracy')
nb_grid.fit(X_train, y_train)

# Logistic Regression
lr_grid = GridSearchCV(lr_model, lr_params, cv=5, scoring='accuracy')
lr_grid.fit(X_train, y_train)

# SVM
svm_grid = GridSearchCV(svm_model, svm_params, cv=5, scoring='accuracy')
svm_grid.fit(X_train, y_train)
```

#### best hyperparameters + scores on test data

```python
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

def evaluate(grid, name):
    best = grid.best_estimator_
    y_pred = best.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    pr, rc, f1, _ = precision_recall_fscore_support(y_test, y_pred, average='binary')
    print(f"{name} | Test Acc: {acc:.4f} Precision: {pr:.4f} Recall: {rc:.4f} F1: {f1:.4f}")

evaluate(nb_grid, "MultinomialNB")
evaluate(lr_grid, "LogisticRegression")
evaluate(svm_grid, "LinearSVC")
```

#### Check for overfitting

```python
def train_test_gap(grid, name):
    best = grid.best_estimator_
    train_acc = accuracy_score(y_train, best.predict(X_train))
    test_acc = accuracy_score(y_test, best.predict(X_test))
    print(f"{name} | Train: {train_acc:.4f} Test: {test_acc:.4f} Gap: {train_acc-test_acc:.4f}")

train_test_gap(nb_grid, "MultinomialNB")
train_test_gap(lr_grid, "LogisticRegression")
train_test_gap(svm_grid, "LinearSVC")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
```

```python
# Define the list of models and their GridSearch objects
grids = [
    ("MultinomialNB", nb_grid),
    ("LogisticRegression", lr_grid),
    ("LinearSVC", svm_grid)
]

# Plot confusion matrices for all models
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, grid) in zip(axes, grids):
    best = grid.best_estimator_
    y_pred = best.predict(X_test)
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Ham","Spam"])
    disp.plot(cmap=plt.cm.Blues, ax=ax, values_format='d')
    ax.set_title(name)

plt.tight_layout()
plt.show()
```
