#### Label Encoder for target

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['income'] = le.fit_transform(df['income'])
df
```

#### Separate Features and Target

```python
X = df.drop('income', axis=1)
y = df['income']
```

#### Identify the column

```python
categorical_cols = X.select_dtypes(include=['object']).columns
categorical_cols
```

#### onehot

#### Define the ColumnTransformer

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(
    transformers=[
        ('onehot', OneHotEncoder(drop='first', sparse_output=False), categorical_cols)
    ],
    remainder='passthrough'
)
```

#### Transform the data

```python
X_encoded = ct.fit_transform(X)
```

#### Convert to DataFrame

```python
import pandas as pd

encoded_feature_names = ct.get_feature_names_out()
X_encoded_df = pd.DataFrame(X_encoded, columns=encoded_feature_names)
```

```python
X_encoded_df
```

#### spilit

```python
# If you haven't already:
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded_df, y,
    test_size=0.2,          # 20% for testing
    random_state=42,        # reproducible
    stratify=y              # optional, good for classification
)
```

#### Train a Tree then a Bagging Model

```python
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
```

#### for a single tree

```python
# Train single decision tree
tree = DecisionTreeClassifier(random_state=42)
tree.fit(X_train, y_train)

```

```python
# Evaluate
y_pred_tree = tree.predict(X_test)
print("Decision Tree Test Accuracy:", accuracy_score(y_test, y_pred_tree))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_tree))
```

```python
y_train_pred = tree.predict(X_train)
print("Tree Training Accuracy:", accuracy_score(y_train, y_train_pred))
```

#### Bagging

```python
bagging_model = BaggingClassifier(
    estimator=DecisionTreeClassifier(),  # base model
    n_estimators=50,                     # number of trees
    random_state=42                     # reproducibility
)
```

```python
bagging_model.fit(X_train, y_train)
```

```python
y_pred = bagging_model.predict(X_test)
```

```python
print("✅ Bagging accuracy test:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
y_train_pred = bagging_model.predict(X_train)
print("Bagging Training Accuracy:", accuracy_score(y_train, y_train_pred))
```

#### Dimensionality reduction with TruncatedSVD

```python
from sklearn.decomposition import TruncatedSVD


svd = TruncatedSVD(n_components=20, random_state=42)
X_train_reduced = svd.fit_transform(X_train)
X_test_reduced = svd.transform(X_test)
```

```python
bagging_model2 = BaggingClassifier(
    estimator=DecisionTreeClassifier(max_depth=3, min_samples_leaf=20),
    n_estimators=100,
    max_samples=0.7,
    max_features=0.6,
    bootstrap=True,
    bootstrap_features=True,  # random subset of features per tree
    oob_score=True,
    random_state=42
)
```

```python
#Train
bagging_model2.fit(X_train_reduced, y_train)
```

```python
#Evaluate
y_train_pred = bagging_model2.predict(X_train_reduced)
y_test_pred = bagging_model2.predict(X_test_reduced)
```

```python

print("Train Accuracy:", accuracy_score(y_train, y_train_pred))
print("Test Accuracy:", accuracy_score(y_test, y_test_pred))
print("OOB Score:", bagging_model2.oob_score_)
print("Confusion Matrix:\n", confusion_matrix(y_test, y_test_pred))
```
