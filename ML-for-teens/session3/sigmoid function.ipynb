#### Step 1: Import Libraries and Create Dataset

```python
import numpy as np
import pandas as pd

# Create a sample customer dataset with Churn
df = pd.DataFrame({
    'Age': [29, 45, 38, 50, 33],
    'Total_Purchase': [5200, 15000, 9800, 20000, 7300],
    'Account_Manager': [0, 1, 0, 1, 0],
    'Years': [2.5, 8.0, 5.5, 10.2, 3.1],
    'Num_Sites': [3, 12, 7, 15, 4],
    'Churn': [1, 0, 1, 0, 1]
})
```

#### Step 2: Define Sigmoid Function

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

#### Step 3: Assign Weights and Bias

```python
w_age = 0.02
w_purchase = 0.001
w_manager = 0.5
w_years = -0.3
w_sites = 0.1
bias = -4.5
```

#### Step 4: Compute Churn Probability for Each Row

```python
def compute_churn_prob(row):
    z = (
        w_age * row['Age'] +
        w_purchase * row['Total_Purchase'] +
        w_manager * row['Account_Manager'] +
        w_years * row['Years'] +
        w_sites * row['Num_Sites'] +
        bias
    )
    return sigmoid(z)

df['Churn_Prob'] = df.apply(compute_churn_prob, axis=1)
```

#### Step 5: Preview the Results

```python
print(df[['Age', 'Total_Purchase', 'Churn_Prob']])
```

#### Step 6: Classify Churn Based on Threshold

```python
def classify_churn(prob):
    return 1 if prob >= 0.5 else 0

df['Churn_Predicted'] = df['Churn_Prob'].apply(classify_churn)
```

#### Step 7: Compare Predictions to Actual Labels

```python
# Compare predicted churn to actual churn
df['Correct'] = df['Churn_Predicted'] == df['Churn']
```
