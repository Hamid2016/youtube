#### Load the Dataset

```python
from sklearn.datasets import load_iris

# Load Iris dataset
X, y = load_iris(return_X_y=True)
```

```python
X
```

#### Split into Train and Test Sets

```python
from sklearn.model_selection import train_test_split

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

#### Train the Gaussian Naive Bayes Model

```python
from sklearn.naive_bayes import GaussianNB

# Create and train the model
model = GaussianNB()
model.fit(X_train, y_train)
```

#### Make Predictions

```python
# Predict on test data
predictions = model.predict(X_test)
print("Predicted classes:", predictions)
```

#### Evaluate Accuracy

```python
from sklearn.metrics import accuracy_score

# Compare predictions to true labels
accuracy = accuracy_score(y_test, predictions)
print("Accuracy:", accuracy)
```

#### View Class Probabilities

```python
# Show predicted probabilities for each class
probs = model.predict_proba(X_test)
print("Class probabilities:\n", probs)
```

#### check overfitting

```python
# Accuracy on training set
train_accuracy = model.score(X_train, y_train)
print("Training Accuracy:", train_accuracy)

# Accuracy on test set
test_accuracy = model.score(X_test, y_test)
print("Test Accuracy:", test_accuracy)
```

#### calculate covariance

```python
from sklearn.datasets import load_iris
import numpy as np
import pandas as pd

# Load full dataset as a dictionary-like object
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Compute covariance matrix between features
cov_matrix = df.cov()

#feature name
feature_names = iris.feature_names
```

```python
df
```

```python


# Display the result
print("Covariance matrix of Iris features:")
print(cov_matrix)
```

#### Covariance Matrix Heatmap

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Plot heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(cov_matrix, annot=True, fmt=".2f",
            xticklabels=feature_names,
            yticklabels=feature_names,
            cmap="coolwarm")
plt.title("Covariance Matrix of Iris Features")
plt.tight_layout()
plt.show()
```
