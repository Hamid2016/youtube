```python
import pandas as pd

# Load the CSV file
df = pd.read_csv('DataSet/WineQT.csv')

# Preview the first few rows
df
```

####  Drop the Id  column

```python
df=df.drop(columns=['Id'])
df
```

#### Split the Data

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=['quality'])
y = df['quality']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

#### Train Three Models

##### 1. Underfitting — Simple Logistic Regression with limited iterations

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

underfit_model = LogisticRegression(max_iter=1)  # Too few iterations
underfit_model.fit(X_train, y_train)

print("Underfitting:")
print("Train Accuracy:", accuracy_score(y_train, underfit_model.predict(X_train)))
print("Test Accuracy:", accuracy_score(y_test, underfit_model.predict(X_test)))
```

##### 2. Appropriate Fit — Logistic Regression with enough iterations

```python
appropriate_model = LogisticRegression(max_iter=100000)
appropriate_model.fit(X_train, y_train)

print("\nAppropriate Fit:")
print("Train Accuracy:", accuracy_score(y_train, appropriate_model.predict(X_train)))
print("Test Accuracy:", accuracy_score(y_test, appropriate_model.predict(X_test)))
```

##### 3. Overfitting — Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

# Initialize and train Decision Tree
model = DecisionTreeClassifier()  # Limit depth for clarity
model.fit(X_train, y_train)

# Evaluate performance
train_acc = accuracy_score(y_train, model.predict(X_train))
test_acc = accuracy_score(y_test, model.predict(X_test))

print(f"Train Accuracy: {train_acc:.2f}")
print(f"Test Accuracy:  {test_acc:.2f}")
```

##### Feature selection 
1.SelectKBest 2.Random Forest 3.LASSO

##### 1.SelectKBest

```python
from sklearn.feature_selection import SelectKBest, mutual_info_classif

selector = SelectKBest(score_func=mutual_info_classif, k=5)
X_selected = selector.fit_transform(X, y)

# Optional: get feature names
selected_feature_names = X.columns[selector.get_support()].tolist()
print("Selected features:", selected_feature_names)
```

##### 2.Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

# Train Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X, y)

# Get feature importances
importances = pd.Series(rf.feature_importances_, index=X.columns)
importances_sorted = importances.sort_values(ascending=False)

# Select top N features (e.g., top 5)
selected_rf = importances_sorted.head(5).index.tolist()
X_rf_selected = X[selected_rf]

print("Top features by Random Forest:")
print(importances_sorted.head(5))
```

##### 3.Lasso

```python
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler
import pandas as pd

# Scale features (important for Lasso)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Fit Lasso model
lasso = Lasso(alpha=0.01)  # You can tune alpha for sparsity
lasso.fit(X_scaled, y)

# Select non-zero coefficients
selected_lasso = X.columns[lasso.coef_ != 0].tolist()
X_lasso_selected = X[selected_lasso]

print("Selected features by LASSO:", selected_lasso)
```

```python
# Find consensus features
common_features = set(selected_feature_names) & set(selected_rf) & set(selected_lasso)
print("✅ Common features across all methods:", list(common_features))
```

#####  common_features selection

```python
new_features = df[list(common_features)]
new_features
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

# Split the data
X_train, X_test, y_train, y_test = train_test_split(new_features, y, test_size=0.3, random_state=42)

# Initialize and train the model
dt_model = DecisionTreeClassifier(
    max_depth=5,              # Limits the tree to 5 levels deep.
    min_samples_split=10,     # A node must have at least 10 samples
    min_samples_leaf=5,       # Each leaf must contain at least 5 samples.
    random_state=42
)

dt_model.fit(X_train, y_train)

# Make predictions
y_pred = dt_model.predict(X_test)

# Evaluate performance
print("✅ Accuracy:", accuracy_score(y_test, y_pred))
print("\n📊 Classification Report:\n", classification_report(y_test, y_pred))
```

```python
# Accuracy on training set
train_preds = dt_model.predict(X_train)
train_acc = accuracy_score(y_train, train_preds)
print("📘 Training Accuracy:", train_acc)

# Accuracy on test set
test_acc = accuracy_score(y_test, y_pred)
print("📕 Test Accuracy:", test_acc)
```
