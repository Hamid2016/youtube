```python
import pandas as pd
df=pd.read_csv("DataSet/Housing.csv")
df.head(5)
```

#### Use one-hot for nominal categorical features

```python
from sklearn.preprocessing import OneHotEncoder

# Define categorical columns
categorical_cols = ['mainroad', 'guestroom', 'basement', 'hotwaterheating',
                    'airconditioning', 'prefarea', 'furnishingstatus']

# One-hot encode
encoder = OneHotEncoder(drop='first', sparse_output=False)
encoded_array = encoder.fit_transform(df[categorical_cols])

# Convert to DataFrame
encoded_df = pd.DataFrame(encoded_array, columns=encoder.get_feature_names_out(categorical_cols))

# Combine with numeric features
df_numeric = df.drop(columns=categorical_cols)
df_encoded = pd.concat([df_numeric.reset_index(drop=True), encoded_df], axis=1)
df_encoded
```

 #### Split Features and Target

```python
X = df_encoded.drop('price', axis=1)
y = df_encoded['price']
```

#### Split the Data

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

#### Linear Regression

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
```

#### Evaluate Overfitting

```python
from sklearn.metrics import r2_score, mean_squared_error

y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

print("R² Train:", r2_score(y_train, y_train_pred))
print("R² Test:", r2_score(y_test, y_test_pred))
print("RMSE Train:", mean_squared_error(y_train, y_train_pred, squared=False))
print("RMSE Test:", mean_squared_error(y_test, y_test_pred, squared=False))
```

#### Add Irrelevant or Noisy Features

```python
import numpy as np

# Add 100 random noise features
np.random.seed(42)
noise = np.random.randn(df_encoded.shape[0], 100)
noise_df = pd.DataFrame(noise, columns=[f'noise_{i}' for i in range(100)])

# Combine with existing features
X_overfit = pd.concat([X, noise_df], axis=1)
```

#### Split noisy data to train and test

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_overfit, y, test_size=0.3, random_state=42)
```

#### Apply linear reg again 

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

model = LinearRegression()
model.fit(X_train, y_train)

y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

print("R² Train:", r2_score(y_train, y_train_pred))
print("R² Test:", r2_score(y_test, y_test_pred))
print("RMSE Train:", mean_squared_error(y_train, y_train_pred, squared=False))
print("RMSE Test:", mean_squared_error(y_test, y_test_pred, squared=False))
```

#### Apply Regularization

##### L1 Regularization (Lasso)

```python
from sklearn.linear_model import Lasso
from sklearn.metrics import r2_score, mean_squared_error

# L1 Regularization
lasso = Lasso()
lasso.fit(X_train, y_train)

# Predictions
y_train_pred_l1 = lasso.predict(X_train)
y_test_pred_l1 = lasso.predict(X_test)

# Evaluation
print("Lasso R² Train:", r2_score(y_train, y_train_pred_l1))
print("Lasso R² Test: ", r2_score(y_test, y_test_pred_l1))
print("Lasso RMSE Train:", mean_squared_error(y_train, y_train_pred_l1, squared=False))
print("Lasso RMSE Test: ", mean_squared_error(y_test, y_test_pred_l1, squared=False))
```

#### L2

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score, mean_squared_error

# Initialize Ridge model with L2 penalty
ridge = Ridge() 
ridge.fit(X_train, y_train)

# Predictions
y_train_pred_l2 = ridge.predict(X_train)
y_test_pred_l2 = ridge.predict(X_test)

# Evaluation
print("Ridge R² Train:", r2_score(y_train, y_train_pred_l2))
print("Ridge R² Test: ", r2_score(y_test, y_test_pred_l2))
print("Ridge RMSE Train:", mean_squared_error(y_train, y_train_pred_l2, squared=False))
print("Ridge RMSE Test: ", mean_squared_error(y_test, y_test_pred_l2, squared=False))
```
