#### k-fold

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import KFold
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import numpy as np
# Load data
digits = load_digits()
X, y = digits.data, digits.target
```

#### Demonstrating Train-Test Split Variability

```python
from sklearn.model_selection import train_test_split
# Run multiple splits to show score variation
for i in range(3):
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=None  # No fixed seed
    )
    
    model = LogisticRegression(max_iter=1000)
    model.fit(X_train, y_train)
    acc = accuracy_score(y_test, model.predict(X_test))
    print(f"Run {i+1} Accuracy:", acc)
```

```python
# Set up K-Fold
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Store fold scores
scores = []

# Loop through folds
for train_idx, test_idx in kf.split(X):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    # Train and evaluate
    model = LogisticRegression(max_iter=1000)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    score = accuracy_score(y_test, y_pred)
    scores.append(score)

# Results
print("K-Fold Accuracy Scores:", scores)
print("Mean Accuracy:", np.mean(scores))
```

#### StratifiedKFold

```python
from sklearn.model_selection import StratifiedKFold

# Set up Stratified K-Fold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Store fold scores
scores = []

# Loop through folds
for fold, (train_idx, test_idx) in enumerate(skf.split(X, y), 1):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    # Train and evaluate
    model = LogisticRegression(max_iter=1000)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    score = accuracy_score(y_test, y_pred)
    scores.append(score)

    print(f"Fold {fold} Accuracy: {score:.4f}")

# Final results
print("\nStratified K-Fold Accuracy Scores:", [f"{s:.4f}" for s in scores])
print("Mean Accuracy:", f"{np.mean(scores):.4f}")
```

#### cross_val_score

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score

# Run cross-validation
model = LogisticRegression()
scores = cross_val_score(model, X, y, cv=5)  # StratifiedKFold used automatically for classification
```
