```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv('Mall_Customers.csv')
```

```python
df
```

```python
df.columns
```

```python
df = df.drop('CustomerID', axis=1)
df
```

```python
X = df

X['Genre'] = X['Genre'].map({'Male': 0, 'Female': 1})

```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
X_scaled = StandardScaler().fit_transform(X)
X_scaled
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

sse = []
K=range(1, 21)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    sse.append(kmeans.inertia_)

plt.plot(K, sse, marker='o')
plt.xlabel('Number of clusters (k)')
plt.ylabel('SSE')
plt.title('Elbow Method for Mall Customers')
plt.grid(True)
plt.show()
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=5, random_state=42)
X['Cluster'] = kmeans.fit_predict(X_scaled)
X
```

```python
df=X
```

```python
df
```

```python
X_features = df.drop('Cluster', axis=1)
y_target = df['Cluster']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_features, y_target, test_size=0.2, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)

from sklearn.metrics import accuracy_score, classification_report
print("Test Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
y_train_pred = knn.predict(X_train)
train_accuracy = accuracy_score(y_train, y_train_pred)
print("Training Accuracy:", train_accuracy)
```

```python
y_test_pred = knn.predict(X_test)
test_accuracy = accuracy_score(y_test, y_test_pred)
print("Test Accuracy:", test_accuracy)
```

```python
from sklearn.neighbors import KNeighborsClassifier

for k in range(3, 16):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    train_acc = knn.score(X_train, y_train)
    test_acc = knn.score(X_test, y_test)
    print(f"k={k}: Train Accuracy={train_acc:.2f}, Test Accuracy={test_acc:.2f}")
```

```python
# Train final KNN model with k=8
knn = KNeighborsClassifier(n_neighbors=8)
knn.fit(X_train, y_train)

# Predict and evaluate
y_pred = knn.predict(X_test)
train_pred = knn.predict(X_train)

# Accuracy scores
print("Training Accuracy:", accuracy_score(y_train, train_pred))
print("Test Accuracy:", accuracy_score(y_test, y_pred))

# Detailed classification report
print(classification_report(y_test, y_pred))
```

```python
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# Split features and target
X_features = df.drop('Cluster', axis=1)
y_target = df['Cluster']
X_train, X_test, y_train, y_test = train_test_split(X_features, y_target, test_size=0.2, random_state=42)

# Define models and hyperparameters
models = {
    'KNN': (KNeighborsClassifier(), {'n_neighbors': [3, 5, 8, 10]}),
    'RandomForest': (RandomForestClassifier(), {'n_estimators': [50, 100], 'max_depth': [None, 10, 20]}),
    'SVM': (SVC(), {'C': [0.1, 1, 10], 'kernel': ['linear', 'rbf']}),
    'LogisticRegression': (LogisticRegression(max_iter=1000), {'C': [0.1, 1, 10]})
}

# Run GridSearchCV with cross-validation
best_models = {}
for name, (model, params) in models.items():
    grid = GridSearchCV(model, params, cv=5, scoring='accuracy')
    grid.fit(X_train, y_train)
    best_models[name] = grid.best_estimator_
    print(f"{name} Best CV Score: {grid.best_score_:.2f}")
    print(f"{name} Best Params: {grid.best_params_}")

# Evaluate best model on test set
for name, model in best_models.items():
    y_pred = model.predict(X_test)
    print(f"\n{name} Test Accuracy: {accuracy_score(y_test, y_pred):.2f}")
    print(classification_report(y_test, y_pred))
```
