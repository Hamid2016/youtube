```python
import pandas as pd

# Sample data
data = {
    'price': [100, 250, 80],
    'quantity': [2, 1, 5],
    'purchase_date': ['2025-11-01', '2025-11-03', '2025-11-05']
}

df = pd.DataFrame(data)
print(df)

```

```python
# Convert purchase_date to datetime
df['purchase_date'] = pd.to_datetime(df['purchase_date'])

# 🎯 Feature Creation Examples

# 1. Total cost
df['total_cost'] = df['price'] * df['quantity']

# 2. Day of week
df['day_of_week'] = df['purchase_date'].dt.day_name()

# 3. Is weekend
df['is_weekend'] = df['day_of_week'].isin(['Saturday', 'Sunday'])

# 4. Price category
df['price_category'] = pd.cut(df['price'], bins=[0, 100, 200, 300], labels=['Low', 'Medium', 'High'])

print(df)
```
