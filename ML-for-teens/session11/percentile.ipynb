```python
import pandas as pd
from scipy.stats import percentileofscore

# Sample scores
scores = [55, 60, 65, 70, 75, 80, 85, 90, 95, 100]

# Create DataFrame
df = pd.DataFrame(scores, columns=['score'])

# Calculate percentile for each row
df['percentile'] = df['score'].apply(lambda x: percentileofscore(df['score'], x))

df
```

#### Detect outliers

```python
import pandas as pd
import numpy as np

# Sample data
scores = [55, 60, 65, 70, 75, 80, 85, 90, 95, 100, 150]  # 150 is an outlier
df = pd.DataFrame(scores, columns=['score'])

# Define percentile thresholds
lower_bound = np.percentile(df['score'], 1)   # 1st percentile
upper_bound = np.percentile(df['score'], 99)  # 99th percentile

# Flag outliers
df['is_outlier'] = (df['score'] < lower_bound) | (df['score'] > upper_bound)

df
```

#### AB_NYC_2019.csv

```python
import pandas as pd
import numpy as np

# Load dataset (after downloading from Kaggle)
df = pd.read_csv('DataSets/AB_NYC_2019.csv')
df
```

```python
# Check for missing values in each column
missing = df.isnull().sum()
missing
```

```python
# Drop all rows with any missing values
df = df.dropna()

# Check shape after dropping
print("Remaining rows:", df.shape[0])
```

```python
df.isnull().sum()
```

#### Detect outliers in the price column using percentiles

```python
# Calculate 1st and 99th percentiles
lower_bound = np.percentile(df['price'], 1)
upper_bound = np.percentile(df['price'], 99)

# Flag outliers
df['is_outlier'] = (df['price'] < lower_bound) | (df['price'] > upper_bound)

# View some results
df[['price', 'is_outlier']]
```

#### Count and inspect outliers

```python
# Count outliers
print("Number of outliers:", df['is_outlier'].sum())

# View outlier rows
outliers = df[df['is_outlier']]
outliers[['price', 'neighbourhood', 'room_type']]
```

#### Dataframe without outliers


```python
clean_df = df[df['is_outlier'] == False]
clean_df
```
