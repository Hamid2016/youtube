```python
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize

# Sample corpus
sentences = [
    "I love deep learning",
    "Word2Vec creates word embeddings",
    "NLP is fun and powerful"
]

# Tokenize sentences
tokenized = [word_tokenize(sentence.lower()) for sentence in sentences]

# Train Word2Vec model
model_w2v = Word2Vec(sentences=tokenized, vector_size=50, window=3, min_count=1, workers=4)

# Get vector for a word
print("Vector for 'learning':\n", model_w2v.wv['learning'])

# Find similar words
print("Most similar to 'nlp':\n", model_w2v.wv.most_similar('nlp'))
```
