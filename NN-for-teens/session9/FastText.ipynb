```python
from gensim.models import FastText
from nltk.tokenize import word_tokenize

# Sample corpus
sentences = [
    "I love deep learning",
    "FastText creates word embeddings",
    "NLP is fun and powerful"
]

# Tokenize
tokenized = [word_tokenize(s.lower()) for s in sentences]
print(tokenized,'tokenized')

# Train FastText model
# FastText model: trains like Word2Vec but also learns subword n‑grams.
# sentences=tokenized → training data
# vector_size=50 → size of each embedding vector
# window=3 → local context window (left/right words)
# min_count=1 → include all words, even rare ones
model_ft = FastText(sentences=tokenized, vector_size=50, window=3, min_count=1)


# Get vector for a word
print("Vector for 'learning':\n", model_ft.wv['learning'])

# Handle unseen word (out-of-vocabulary)
print("Vector for 'learnings':\n", model_ft.wv['learnings'])
```

```python
sim = model_ft.wv.similarity('learning', 'learnings')
print(sim)
```

```python
print(model_ft.wv.most_similar("learning"))
```
