```python
import numpy as np

# Load GloVe embeddings (e.g., glove.6B.50d.txt)
#glove.6B.50d.txt is one of the pre‑trained GloVe embedding files released by Stanford.
embeddings_index = {}
with open("Datasets\glove.6B.50d\glove.6B.50d.txt", encoding="utf8") as f: 
    for line in f:
        values = line.split()
        word = values[0]
        vector = np.asarray(values[1:], dtype='float32')
        embeddings_index[word] = vector

# Get vector for a word
print("Vector for 'learning':\n", embeddings_index['learning'])

# Compare similarity (cosine similarity)
from numpy.linalg import norm

def cosine_similarity(vec1, vec2):
    return np.dot(vec1, vec2) / (norm(vec1) * norm(vec2))

print("Similarity between 'king' and 'queen':", cosine_similarity(embeddings_index['king'], embeddings_index['queen']))
```
