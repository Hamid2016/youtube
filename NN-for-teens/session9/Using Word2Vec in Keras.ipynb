```python
from gensim.models import Word2Vec
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense
from nltk.tokenize import word_tokenize

# Sample corpus
sentences = ["I love deep learning", "Word2Vec creates word embeddings", "NLP is fun"]

# Tokenize
tokenized = [word_tokenize(s.lower()) for s in sentences]

# Train Word2Vec
w2v_model = Word2Vec(sentences=tokenized, vector_size=50, window=3, min_count=1)

# Build embedding matrix
vocab = list(w2v_model.wv.key_to_index.keys())
embedding_matrix = [w2v_model.wv[word] for word in vocab]

# Keras model
model = Sequential([
    Embedding(input_dim=len(vocab), output_dim=50, weights=[embedding_matrix], trainable=False),
    GlobalAveragePooling1D(),
    Dense(1, activation='sigmoid')
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```
