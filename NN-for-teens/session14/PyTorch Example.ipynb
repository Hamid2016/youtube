#### Import PyTorch and the tools we need.

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

#### Make fake input data: 16 rows, each with 10 numbers.

```python
x = torch.randn(16, 10)
```

#### Make fake target values: 16 rows, each with 1 number.

```python
y = torch.randn(16, 1)
```

Build a tiny model:

First layer: 10 → 32

ReLU

Output layer: 32 → 1

```python
model = nn.Sequential(
    nn.Linear(10, 32),
    nn.ReLU(),
    nn.Linear(32, 1)
)
```

#### Choose loss function and optimizer.

```python
loss_fn = nn.MSELoss()
opt = optim.Adam(model.parameters(), lr=0.001)
```

#### Model makes predictions.

```python
pred = model(x)
```

#### Compare prediction with target.

```python
loss = loss_fn(pred, y)
```

#### Compute gradients.

```python
loss.backward()

```

#### Update model weights.

```python
print("PyTorch loss:", loss.item())
```
