```python
import numpy as np

# Simple quadratic loss: f(w) = (w - 3)^2
def loss(w, x):
    return (w - x)**2

def grad(w, x):
    return 2 * (w - x)

# Dataset (pretend these are samples)
data = np.array([3, 3, 3, 3, 3])  # all centered at 3 for simplicity
w_init = 0.0
lr = 0.1
epochs = 2

# -------------------------------
# 1. Batch Gradient Descent
# -------------------------------
w = w_init
print("Batch Gradient Descent:")
for epoch in range(epochs):
    # gradient over all samples
    g = np.mean([grad(w, x) for x in data])
    w -= lr * g
    print(f"Epoch {epoch+1}, w = {w:.4f}")

# -------------------------------
# 2. Stochastic Gradient Descent
# -------------------------------
w = w_init
print("\nStochastic Gradient Descent:")
for epoch in range(epochs):
    for x in data:  # update per sample
        g = grad(w, x)
        w -= lr * g
        print(f"Epoch {epoch+1}, sample {x}, w = {w:.4f}")

# -------------------------------
# 3. Mini-Batch Gradient Descent
# -------------------------------
batch_size = 2
w = w_init
print("\nMini-Batch Gradient Descent:")
for epoch in range(epochs):
    # shuffle data
    np.random.shuffle(data)
    # split into batches
    for i in range(0, len(data), batch_size):
        batch = data[i:i+batch_size]
        g = np.mean([grad(w, x) for x in batch])
        w -= lr * g
        print(f"Epoch {epoch+1}, batch {batch}, w = {w:.4f}")
```
