```python
from gensim.models import Word2Vec

# Sample corpus (list of tokenized sentences)
sentences = [
    ["i", "love", "machine", "learning"],
    ["word2vec", "creates", "word", "embeddings"],
    ["embeddings", "capture", "semantic", "relationships"],
    ["deep", "learning", "models", "use", "embeddings"]
]

# Train Word2Vec model
model = Word2Vec(
    sentences,
    vector_size=100,   # embedding dimension, each word becomes a 100-dim vector
    window=5,          # context window, look 5 words lest/right
    min_count=1,       # ignore words with freq < 1, keep allnwords
    workers=4,         # CPU cores, threads
    sg=1               # 1 = skip-gram, 0 = CBOW
)

# Get vector for a word,
print(model.wv["embeddings"])

# Find similar words, from most to least relevant
print(model.wv.most_similar("learning"))
```

### Your dataset is tiny — only 4 sentences.
Word2Vec needs millions of sentences to learn strong semantic structure.
