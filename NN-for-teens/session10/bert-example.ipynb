```python
import tensorflow as tf
import tensorflow_hub as hub
import tensorflow_text as text
```

```python
preprocess_url = "https://tfhub.dev/tensorflow/bert_en_uncased_preprocess/3"
encoder_url = "https://tfhub.dev/tensorflow/small_bert/bert_en_uncased_L-12_H-768_A-12/2"
```

```python
preprocess = hub.KerasLayer(preprocess_url)
encoder = hub.KerasLayer(encoder_url, trainable=False)
```

```python
text_input = ["this is a simple test sentence","I love data science"]
tokens = preprocess(text_input)
tokens.keys()
```

```python
tokens.keys()
```

# input_type_ids are segment IDs.
It tells BERT which sentence each token belongs to inside a single input example.

0 → Sentence A (segment A)

1 → Sentence B (segment B)

```python
tokens['input_type_ids']
```

# input_mask tells BERT which positions are real tokens and which positions are padding.

1 → real token (word, subword, CLS, SEP)

0 → padding (just filler to reach max length)

```python
tokens['input_mask']
```

# Also called segment IDs or token type IDs.
It is the sequence of token IDs produced by the BERT tokenizer.

These IDs correspond to:

[CLS] token

WordPiece tokens

[SEP] token

Padding (0)

```python
tokens['input_word_ids']
```

```python
outputs = encoder(tokens)
outputs.keys()
```

1. pooled_output
Shape: (batch_size, 768)

The CLS sentence embedding

Used for sentence‑level tasks (classification, sentiment)

2. sequence_output
Shape: (batch_size, seq_len, 768)

The embedding for every token

Used for token‑level tasks (NER, QA)

3. encoder_outputs
A list of all hidden layers inside BERT
(12 layers for BERT‑base)

Used when you want layer‑by‑layer representations

4. default
Same as pooled_output

Added for Keras compatibility

```python


embedding = outputs["pooled_output"]
print(embedding.shape)
print(embedding)
```
