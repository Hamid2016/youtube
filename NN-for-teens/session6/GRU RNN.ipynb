```python
import tensorflow as tf
import numpy as np

# ----- Fake real‑world sequence -----
sequence = np.array([
    [0.1, 0.2, 0.3],   # "Hey"
    [0.5, 0.1, 0.0],   # "I"
    [0.3, 0.8, 0.2],   # "need"
    [0.9, 0.4, 0.1]    # "help"
], dtype=np.float32)

sequence = sequence.reshape(1, 4, 3)  # (batch, timesteps, features)

# ----- GRU -----
gru = tf.keras.layers.GRU(
    units=4,
    return_sequences=True,   # show hidden state at each step
    return_state=False
)

outputs = gru(sequence)

print("GRU hidden state at each time step:\n")
for t, h in enumerate(outputs[0]):
    print(f"t={t+1} -> {h.numpy()}")
```
