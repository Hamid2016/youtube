```python
import tensorflow as tf
import pandas as pd
import numpy as np

# -----------------------------
# 1. Load CSV dataset
# -----------------------------
df = pd.read_csv("mini_sentiment.csv")

texts = df["text"].tolist()
labels = df["label"].values

# -----------------------------
# 2. Vectorize text
# -----------------------------
vectorizer = tf.keras.layers.TextVectorization(
    max_tokens=1000,
    output_sequence_length=20
)

vectorizer.adapt(texts)
X = vectorizer(texts)

# -----------------------------
# 3. Train / Val / Test split
# -----------------------------
X_train, y_train = X[:6], labels[:6]
X_val,   y_val   = X[6:8], labels[6:8]
X_test,  y_test  = X[8:], labels[8:]

# -----------------------------
# 4. Build + train + evaluate
# -----------------------------
def run_model(cell):
    model = tf.keras.Sequential([
        tf.keras.layers.Embedding(1000, 16),
        cell(16),
        tf.keras.layers.Dense(1, activation="sigmoid")
    ])

    model.compile(
        loss="binary_crossentropy",
        optimizer="adam",
        metrics=["accuracy"]
    )

    print("\nTraining:", cell.__name__)
    model.fit(X_train, y_train, validation_data=(X_val, y_val),
              epochs=5, verbose=1)

    loss, acc = model.evaluate(X_test, y_test, verbose=0)
    print("Test Accuracy:", acc)

# -----------------------------
# 5. Run all 3 models
# -----------------------------
run_model(tf.keras.layers.SimpleRNN)
run_model(tf.keras.layers.GRU)
run_model(tf.keras.layers.LSTM)
```
