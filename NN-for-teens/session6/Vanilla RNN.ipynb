```python
import tensorflow as tf
import numpy as np

# ----- Fake real‑world sequence -----
# Example sentence: ["Hey", "I", "need", "help"]
# We'll represent each word with a simple 3‑dim vector
sequence = np.array([
    [0.1, 0.2, 0.3],   # "Hey"
    [0.5, 0.1, 0.0],   # "I"
    [0.3, 0.8, 0.2],   # "need"
    [0.9, 0.4, 0.1]    # "help"
], dtype=np.float32)

sequence = sequence.reshape(1, 4, 3)  # (batch, timesteps, features)

# ----- Vanilla RNN -----
rnn = tf.keras.layers.SimpleRNN(
    units=4,
    return_sequences=True,   # so we can see hidden state at each step
    return_state=False # do NOT return the final hidden state separately
)

outputs = rnn(sequence)

print("Hidden state at each time step:\n")
for t, h in enumerate(outputs[0]):
    print(f"t={t+1} -> {h.numpy()}")
```
