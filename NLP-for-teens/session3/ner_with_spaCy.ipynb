```python
import spacy   # import spaCy

nlp = spacy.load("en_core_web_sm")   # load English model

text = "Alice went to Germany in 2024 to join Google."   # sample text

doc = nlp(text)   # process text

for ent in doc.ents:   # loop through detected entities
    print(ent.text, ent.label_)   # print entity + its type
```

```python
# print all NER labels in the model
print(nlp.get_pipe("ner").labels)
```

```python
print(spacy.explain("GPE"))     # explain NER tag
```
