```python
import spacy   # import the spaCy library

nlp = spacy.load("en_core_web_sm")   # load the small English model

text = "She quickly ran to the trees."   # the sentence we want to analyze

doc = nlp(text)   # process the text with spaCy

for token in doc:   # loop through each token (word)
    print(token.text, token.pos_)   # print the word and its POS tag
```

ADJ — adjective; describes a noun
big, fast, beautiful

ADP — adposition; prepositions and postpositions
to, in, on, under

ADV — adverb; modifies verbs/adjectives
quickly, very, silently

AUX — auxiliary verb; helps main verbs
is, have, will

CONJ — conjunction; connects words/phrases
and, but, or

CCONJ — coordinating conjunction
and, but, or

DET — determiner; introduces nouns
the, a, this

INTJ — interjection; emotional expressions
wow, oh, hey

NOUN — noun; person, place, thing
tree, car, idea

PROPN — proper noun; specific names
London, Alice, Google

NUM — number
one, five, 2024

PART — particle; small function words
not, ’s

PRON — pronoun; replaces nouns
she, they, it

PUNCT — punctuation
., !, ?

SCONJ — subordinating conjunction
because, although, if

SYM — symbol
%, $, +

VERB — verb; action or state
run, think, be

X — other; unknown or foreign words
emoji, foreign fragments
