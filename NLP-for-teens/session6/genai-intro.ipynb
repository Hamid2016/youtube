```python
from transformers import pipeline
# import the pipeline function from Hugging Face

# Create a text-generation pipeline (GenAI model)
generator = pipeline("text-generation", model="gpt2")
# load a GPT‑2 text‑generation model for producing new text

# Prompt
prompt = "In the future, artificial intelligence will"
# the starting sentence the model will continue

# Generate text
result = generator(
    prompt,
    max_length=60,  # limit total tokens (prompt + generated text) to 60
    # controls how long the final output can be

    num_return_sequences=1,  # generate only one output sequence
    # if >1, it would return multiple different generations

    do_sample=True,  # enable randomness instead of deterministic output
    # allows creative, varied text instead of always the same result

    temperature=0.8,  # control randomness strength (0.8 = balanced creativity)
    # higher values = more unusual word choices; lower = safer output
)

print(result[0]["generated_text"])
# print the generated text from the first (and only) sequence
```
