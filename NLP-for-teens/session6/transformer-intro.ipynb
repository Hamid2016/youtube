```python
from transformers import pipeline
# import the pipeline function from Hugging Face

classifier = pipeline("sentiment-analysis")
# load a ready sentiment‑analysis model (default: DistilBERT SST‑2)

text = "I really enjoyed learning about transformers today!"
# the input sentence we want to classify

result = classifier(text)
# run the classifier and get the label + confidence score

print(result)
# show the output (POSITIVE or NEGATIVE with probability)
```
