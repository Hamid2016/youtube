#### Import the library

```python
from transformers import pipeline
```

#### Create the pipeline

```python
sentiment_model = pipeline("sentiment-analysis")
```

#### Sends the text into the sentiment‑analysis model to get a prediction

```python
result = sentiment_model("I love learning Transformers!")
```

```python
result
```

```python
r1=sentiment_model("I hate learning Transformers!")
r1
```

# This output means the model thinks the sentence is positive with very high confidence.
