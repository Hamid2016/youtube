```python
from nltk.stem import PorterStemmer  # Import the stemmer
from nltk.tokenize import word_tokenize  # Import tokenizer
import nltk  # NLTK main package

nltk.download('punkt')  # Download tokenizer data

ps = PorterStemmer()  # Create stemmer object

text = "running runs easily fairly"  # Input text
words = word_tokenize(text)  # Split text into words

stems = [ps.stem(w) for w in words]  # Apply stemming to each word
print(stems)  # Show stemmed words
```
