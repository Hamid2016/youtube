```python
import spacy

# Load spaCy's English model
nlp = spacy.load("en_core_web_sm")

# Example text for the pipeline
text = "This is an example sentence to demonstrate an NLP pipeline in spaCy."

# Process the text through the pipeline
doc = nlp(text)

# Extract and print tokens and lemmas
for token in doc:
    print(f"Token: {token.text}, Lemma: {token.lemma_}")
```
