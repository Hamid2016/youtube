```python
import spacy  # Import spaCy

nlp = spacy.load("en_core_web_sm")  # Load English model

text = "running runs easily fairly"  # Input text
doc = nlp(text)  # Process text with spaCy pipeline

lemmas = [token.lemma_ for token in doc]  # Extract lemmas
print(lemmas)  # Show lemmatized words
```
