# Import libraries

```python
import pandas as pd
import numpy as np
```

# Load dataset

```python
rows = 2000  # set None to load full dataset

data = pd.read_csv('/kaggle/input/datasets/rtatman/blog-authorship-corpus/blogtext.csv', nrows=rows)

r, c = data.shape
print(f'There are {r} rows and {c} columns')
```

```python
data
```

```python
text_col = data['text']
text_col[:5]
text_col
```

# Simple preprocessing

```python
text_clean = text_col.str.lower()
text_clean[:5]
text_clean
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=500)
vecs = tfidf.fit_transform(text_clean)

vecs.shape
```

# Show sample vectors

```python
vecs.toarray()[:3]
```

# Count categories in each column

```python
print("Gender categories:", data['gender'].nunique())
print("Age categories:", data['age'].nunique())
print("Topic categories:", data['topic'].nunique())
print("Blogger ID categories:", data['id'].nunique())
print("Sign categories:", data['sign'].nunique())
```

# Add TF‑IDF as one column

```python
import numpy as np

data['tfidf'] = list(vecs.toarray())
data
```

```python
data.columns
```

# Keep only tfidf features and gender label for classification

```python
new_data = data[['tfidf', 'gender']]
new_data
```

# Check gender balance

```python
# check if male and female counts are balanced
print(new_data['gender'].value_counts())
```

#  Convert gender labels to numeric values

```python
gender_map = {'male': 0, 'female': 1}
y = new_data['gender'].map(gender_map)
```

# Define X as the tfidf feature matrix

```python
X = np.array(new_data['tfidf'].tolist())
```

# Define all classifiers once


```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB

log_reg = LogisticRegression(max_iter=1000)
svm = SVC()
rf = RandomForestClassifier()
nb = GaussianNB()
```

# Define all hyperparameters once


```python
log_reg_params = {
    "C": [0.1, 1, 10],
    "solver": ["lbfgs"]
}

svm_params = {
    "C": [0.1, 1, 10],
    "kernel": ["linear", "rbf"]
}

rf_params = {
    "n_estimators": [100, 200],
    "max_depth": [None, 10, 20]
}

nb_params = {
    "var_smoothing": [1e-9, 1e-8, 1e-7]
}
```

# Train each classifier with GridSearchCV

```python
from sklearn.model_selection import GridSearchCV

log_reg_grid = GridSearchCV(log_reg, log_reg_params, cv=5, scoring="accuracy")
log_reg_grid.fit(X, y)

svm_grid = GridSearchCV(svm, svm_params, cv=5, scoring="accuracy")
svm_grid.fit(X, y)

rf_grid = GridSearchCV(rf, rf_params, cv=5, scoring="accuracy")
rf_grid.fit(X, y)

nb_grid = GridSearchCV(nb, nb_params, cv=5, scoring="accuracy")
nb_grid.fit(X, y)
```

# Collect results

```python
results = {
    "log_reg": {
        "score": log_reg_grid.best_score_,
        "params": log_reg_grid.best_params_
    },
    "svm": {
        "score": svm_grid.best_score_,
        "params": svm_grid.best_params_
    },
    "rf": {
        "score": rf_grid.best_score_,
        "params": rf_grid.best_params_
    },
    "nb": {
        "score": nb_grid.best_score_,
        "params": nb_grid.best_params_
    }
}

results
```

# View results cleanly (sorted by score)

```python
sorted_results = sorted(
    results.items(),
    key=lambda x: x[1]["score"],
    reverse=True
)

sorted_results
```
