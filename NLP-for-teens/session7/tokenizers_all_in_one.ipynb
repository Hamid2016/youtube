# 1. Load Tokenizers

```python
# Load three tokenizers: WordPiece (BERT), BPE (GPT-2), SentencePiece (T5)
from transformers import BertTokenizer, GPT2Tokenizer, T5Tokenizer

bert_tok = BertTokenizer.from_pretrained("bert-base-uncased")     # WordPiece
gpt2_tok = GPT2Tokenizer.from_pretrained("gpt2")                  # BPE
t5_tok = T5Tokenizer.from_pretrained("t5-small")                  # SentencePiece
```

# 2. Input Text

```python
# Sample text for testing all tokenizers
text = "Transformers changed NLP forever."
```

# 3. WordPiece (BERT)

```python
# WordPiece splits words into subwords using "##"
wp_tokens = bert_tok.tokenize(text)
print("WordPiece:", wp_tokens)
```

# 4. BPE (GPT‑2)

```python
# BPE merges most frequent character pairs to create subwords
bpe_tokens = gpt2_tok.tokenize(text)
print("BPE:", bpe_tokens)
```

# 5. SentencePiece (T5)

```python
# SentencePiece learns tokens from raw text and uses "▁" to mark spaces
sp_tokens = t5_tok.tokenize(text)
print("SentencePiece:", sp_tokens)
```

# 6. Encode → IDs

```python
# Convert tokens into integer IDs (model input)
print("WP IDs:", bert_tok.encode(text))
print("BPE IDs:", gpt2_tok.encode(text))
print("SP IDs:", t5_tok.encode(text))
```

# 7. Decode → Back to Text

```python
# Convert IDs back to text
print("WP Decode:", bert_tok.decode(bert_tok.encode(text)))
print("BPE Decode:", gpt2_tok.decode(gpt2_tok.encode(text)))
print("SP Decode:", t5_tok.decode(t5_tok.encode(text)))
```

# 8. Compare All Tokenizers

```python
# Simple comparison function for quick testing
def compare_all(txt):
    print("Text:", txt)
    print("WordPiece:", bert_tok.tokenize(txt))
    print("BPE:", gpt2_tok.tokenize(txt))
    print("SentencePiece:", t5_tok.tokenize(txt))

compare_all("Transformers changed NLP forever.")
```
