```python
# Install + load
!pip install transformers

from transformers import BartTokenizer

sentence = "The model learns patterns from large amounts of text."

# --- BART tokenizer ---
bart = BartTokenizer.from_pretrained("facebook/bart-base")

# Tokens
print("BART tokens:", bart.tokenize(sentence))

# Encode IDs
bart_ids = bart.encode(sentence)
print("BART IDs:", bart_ids)

# Decode
print("BART decode:", bart.decode(bart_ids))
```
