# Write the server:

```python
import nest_asyncio                      # import nest_asyncio for running async code in notebooks
nest_asyncio.apply()                     # apply it so asyncio works inside Jupyter

with open("server.py", "w") as f:        # open a file called server.py for writing
    f.write('''                          # start writing the server code as a string
from mcp.server.fastmcp import FastMCP   # import FastMCP, the helper for building MCP servers

mcp = FastMCP("demo")                    # create an MCP server named "demo"

@mcp.tool()                              # register the next function as a tool
def greet(name: str) -> str:             # define the tool function, takes a name, returns text
    """Say hello to someone."""          # docstring tells the AI what this tool does
    return f"Hello, {name}!"             # return the greeting

if __name__ == "__main__":               # only run when the file is executed directly
    mcp.run()                            # start the MCP server
''')                                     # end of the server code string
```

#  Run the client

```python
import asyncio                                                    # import asyncio for async code
from mcp import ClientSession, StdioServerParameters              # import client and server params
from mcp.client.stdio import stdio_client                         # import stdio transport for connecting

async def main():                                                 # define async main function
    params = StdioServerParameters(command="python", args=["server.py"])  # tell client how to start the server
    async with stdio_client(params) as (read, write):            # start the server and open a connection
        async with ClientSession(read, write) as session:         # create a client session over the connection
            await session.initialize()                            # initialize the session with the server
            result = await session.call_tool("greet", {"name": "Sara"})  # call the greet tool with name "Sara"
            print(result.content[0].text)                         # print the text the tool returned

asyncio.run(main())                                               # run the async main function
```
