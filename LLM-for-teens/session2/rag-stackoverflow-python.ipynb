# step 1 – import

```python
# import tools
import pandas as pd
from sentence_transformers import SentenceTransformer, util
```

# step 2 – load model

```python
# load small embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")
```

# step 3 – load dataset

```python
# load Kaggle dataset (StackOverflow Python Questions)
df = pd.read_csv("/kaggle/input/datasets/stackoverflow/pythonquestions/Questions.csv",
                 encoding="latin-1")
df.head()
```

# step 4 — load answers

```python
# load answers file
ans = pd.read_csv("/kaggle/input/datasets/stackoverflow/pythonquestions/Answers.csv",
                  encoding="latin-1")
ans
```

# step 5 — merge questions + answers

```python
# connect questions to answers
merged = pd.merge(df, ans, left_on="Id", right_on="ParentId", how="inner")
merged
```

# step 6 — take text columns

```python
# question titles and answer bodies
questions = merged["Title"].astype(str).tolist()
answers   = merged["Body_y"].astype(str).tolist()
```

# step 7 — embeddings

```python
q_vecs = model.encode(questions)
```

# step 8 — query

```python
query = "how to fix login error"
query_vec = model.encode(query)
```

# step 9 — similarity

```python
scores = util.cos_sim(query_vec, q_vecs)
top_id = scores.argmax().item()
```

# step 10 — return matched Q&A

```python
print("question:", questions[top_id])
print("answer:", answers[top_id])
```
