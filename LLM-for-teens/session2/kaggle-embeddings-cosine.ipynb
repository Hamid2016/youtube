# main code

```python
# import model + similarity utility
from sentence_transformers import SentenceTransformer, util

# load small embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# text 1
emb1 = model.encode("king")

# text 2
emb2 = model.encode("queen")

# cosine similarity
print(util.cos_sim(emb1, emb2))
```
